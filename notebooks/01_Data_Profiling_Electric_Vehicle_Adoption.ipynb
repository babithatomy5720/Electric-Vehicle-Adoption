```
#Missing values
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df) * 100).round(2),
    "Data_Type": df.dtypes
})

missing_summary.sort_values(
    by="Missing_Percentage",
    ascending=False
)
```

```
#duplicate records
duplicate_count = df.duplicated().sum()

print(f"Duplicate Records: {duplicate_count:,}")
print(f"Duplicate Percentage: {(duplicate_count / len(df) * 100):.2f}%")
```

```
#unique values
unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": [df[col].nunique() for col in df.columns],
    "Missing_Values": [df[col].isnull().sum() for col in df.columns],
    "Data_Type": [df[col].dtype for col in df.columns]
})

unique_summary
```

```
# ============================================================
# STEP 3: BUSINESS-CRITICAL DATA QUALITY CHECKS
# ============================================================

# 1. CAFV Eligibility Distribution
print("=" * 70)
print("1. CAFV ELIGIBILITY DISTRIBUTION")
print("=" * 70)

cafv_col = "Clean Alternative Fuel Vehicle (CAFV) Eligibility"

print(df[cafv_col].value_counts(dropna=False))
print("\nPercentage Distribution:")
print((df[cafv_col].value_counts(normalize=True, dropna=False) * 100).round(2))


# 2. Electric Vehicle Type Distribution
print("\n" + "=" * 70)
print("2. ELECTRIC VEHICLE TYPE DISTRIBUTION")
print("=" * 70)

ev_type_col = "Electric Vehicle Type"

print(df[ev_type_col].value_counts(dropna=False))
print("\nPercentage Distribution:")
print((df[ev_type_col].value_counts(normalize=True, dropna=False) * 100).round(2))


# 3. Electric Range Analysis
print("\n" + "=" * 70)
print("3. ELECTRIC RANGE ANALYSIS")
print("=" * 70)

print(df["Electric Range"].describe())

zero_range_count = (df["Electric Range"] == 0).sum()
missing_range_count = df["Electric Range"].isnull().sum()

print(f"\nRecords with zero electric range: {zero_range_count:,}")
print(f"Records with missing electric range: {missing_range_count:,}")

print(
    f"Zero range percentage: "
    f"{zero_range_count / len(df) * 100:.2f}%"
)

print(
    f"Missing range percentage: "
    f"{missing_range_count / len(df) * 100:.2f}%"
)


# 4. Model Year Analysis
print("\n" + "=" * 70)
print("4. MODEL YEAR ANALYSIS")
print("=" * 70)

print(df["Model Year"].describe())

print("\nModel Year Distribution:")
print(df["Model Year"].value_counts().sort_index())


# 5. Geographic Coverage
print("\n" + "=" * 70)
print("5. GEOGRAPHIC COVERAGE")
print("=" * 70)

print(f"Unique Counties: {df['County'].nunique():,}")
print(f"Unique Cities: {df['City'].nunique():,}")
print(f"Unique States: {df['State'].nunique():,}")

print("\nTop 10 Counties:")
print(df["County"].value_counts().head(10))

print("\nTop 10 Cities:")
print(df["City"].value_counts().head(10))
```

```
data_quality_report = pd.DataFrame({
    "Feature": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2).values,
    "Unique_Values": [
        df[col].nunique() for col in df.columns
    ]
})

data_quality_report
```

```
# Duplicate records
print("Duplicate Records:", df.duplicated().sum())

# CAFV Eligibility
print("\nCAFV Eligibility:")
display(
    df[
        "Clean Alternative Fuel Vehicle (CAFV) Eligibility"
    ].value_counts(dropna=False)
)

# Electric Vehicle Type
print("\nElectric Vehicle Type:")
display(
    df["Electric Vehicle Type"]
    .value_counts(dropna=False)
)

# Electric Range
print("\nElectric Range Summary:")
display(df["Electric Range"].describe())

print(
    "\nZero Electric Range Records:",
    (df["Electric Range"] == 0).sum()
)
```

```
data_quality_report.to_csv(
    "data_quality_report.csv",
    index=False
)
```
